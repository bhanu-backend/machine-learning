```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("insurance_data.csv")
df
```

```python
plt.scatter(df.age , df.bought_insurance , marker = "+" ,color = "red"  )
```

```python
X = df[['age']]
y = df['bought_insurance']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.5)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
model.fit(X_train , y_train)
```

```python
model.predict(X_test)
```

```python
y_test
```

```python
model.score(X_test,y_test)
```

```python
model.predict_proba(X_test)
```

```python
from sklearn.datasets import load_digits
import matplotlib.pyplot as plt
```

```python
digits = load_digits()
dir(digits)
```

```python
digits.data[0]
```

```python
digits.images[0]
```

```python
digits.target[0:5]
```

```python
plt.gray()
plt.matshow(digits['images'][0])
```

```python
plt.gray()

for i in range(10):
    plt.matshow(digits['images'][i])
    
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(digits['data'], digits['target'], test_size=0.33)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.predict(X_test)
```

```python
model.score(X_test , y_test)
```

```python
model.predict([digits['data'][999]])
```

```python
plt.gray()
plt.matshow(digits['images'][999])
```

```python
from sklearn.metrics import confusion_matrix
y_predicted = model.predict(X_test)
matrixx = confusion_matrix(y_test,y_predicted)
```

```python
matrixx
```

```python
import seaborn as sns
sns.heatmap(matrixx, annot=True)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.show()
```
